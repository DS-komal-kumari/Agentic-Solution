```python
# Initialize OpenAI client to interact with Databricks Foundation Model API
# This cell demonstrates basic LLM interaction using the OpenAI SDK

from openai import OpenAI
import os

# How to get your Databricks token: https://docs.databricks.com/en/dev-tools/auth/pat.html
# DABRICKS_TOKEN = os.environ.get('DATABRICKS_TOKEN')
# Alternatively in a Databricks notebook you can use this:
DATABRICKS_TOKEN = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

client = OpenAI(
    api_key=DATABRICKS_TOKEN,
    base_url="https://dbc-ed5fac1e-f136.cloud.databricks.com/serving-endpoints"
)

response = client.chat.completions.create(
    model="databricks-gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": "What is an LLM agent?"
        }
    ],
    max_tokens=5000
)

print(response.choices[0].message.content)
```

```python
# Extract and print specific text content from the response
print(response.choices[0].message.content[1].get('text'))
```


An **LLM (Large Language Model) agent** is a software system that couples a powerful language model—such as GPT‑4, Claude, LLaMA, or Gemini—with additional reasoning, memory, and interaction mechanisms so that it can act autonomously or semi‑autonomously to achieve a user‑specified goal.  

In other words, the LLM is the “brain” that understands and generates natural‑language text, while the surrounding **agent framework** supplies the “body” that lets the model:

1. **Perceive** external data (e.g., web pages, APIs, sensor feeds).  
2. **Plan & reason** about a sequence of actions required to solve a problem.  
3. **Execute** those actions (calling APIs, manipulating files, sending messages, etc.).  
4. **Maintain state** across turns (memory, context windows, or external databases).  
5. **Self‑evaluate** results and iterate until the goal is satisfied or a stopping condition is met
